# Datlas — exploration des données Twitch

Analyse de 7 mois de statistiques Twitch (juin → décembre 2025) : croissance, régularité, rétention.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

p = Path('data/twitch_monthly.csv')
if not p.exists():
    p = Path('../data/twitch_monthly.csv')
df = pd.read_csv(p, parse_dates=['month']).sort_values('month')
df
```

## Vue d'ensemble

```
print('Heures streamees :', round(df.hours_streamed.sum(), 1))
print('Heures regardees :', round(df.hours_watched.sum(), 1))
print('Nouveaux followers :', int(df.new_followers.sum()))
print('Revenus (EUR) :', round(df.revenue_eur.sum(), 2))
df.describe().round(2)
```

## Régularité de stream vs croissance de followers

```
fig, ax1 = plt.subplots(figsize=(9, 4.5))
m = df.month.dt.strftime('%b')
ax1.bar(m, df.hours_streamed, color='#185FA5', label='Heures streamees')
ax1.set_ylabel('Heures streamees')
ax2 = ax1.twinx()
ax2.plot(m, df.new_followers, color='#D85A30', marker='o', linewidth=2.5, label='Nouveaux followers')
ax2.set_ylabel('Nouveaux followers')
plt.title('Plus je streame, plus je grandis - sauf en septembre')
fig.tight_layout(); plt.show()
```

## Lecture

- **La régularité est le premier levier** : juillet-août (volumes max) = meilleure croissance ; l'arrêt de septembre casse l'élan, jamais totalement retrouvé ensuite.
- **Bonne rétention** : ~3 spectateurs restent des heures — le contenu tient, c'est la découverte qui pèche.
